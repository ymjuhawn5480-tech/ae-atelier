| fuel | distance |
| ---- | ------- |
| 4    | 6       |

```python
#연료에 대한 자동차의 이동 거리 데이터
car = [[ 4,  6],
       [ 8,  5],
       [12, 15],
       [16, 12],
       [20, 20]]
```

### 연료에 대한 이동 거리 관계 시각화

```python
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize'] = [5, 3] # [width, height] (inches)
```

```python
#연료에 대한 자동차의 이동 거리 데이터
car = [[ 4,  6],
       [ 8,  5],
       [12, 15],
       [16, 12],
       [20, 20]]

#데이터 추출
X = []
y = []
for i in car :
  X.append(i[0]) #? 작성
  y.append(i[1]) #? 작성

print(X)
print(y)
```

```python
plt.scatter(X, y) #? 작성
plt.xlabel("fuel")
plt.ylabel("distance")
plt.show()
```

### 연료에 따른 자동차의 이동 거리 예측 모델, 가중치 찾기

예측선(Prediction Line), 예측 모델 y=aX (a는 기울기)

```python
a = 0.5
y_pred = [i*a for i in X] #y_pred=aX
print(y_pred)
```

```python
plt.scatter(X, y) #실측값
plt.plot(X, y_pred, "r")  #예측값, ? 작성
plt.xlabel("fuel")
plt.ylabel("distance")
plt.show()
```

```python
#데이터를 잘 설명할 수 있는 a값을 찾아보자.

a = 0.9         #? 작성
y_pred = [i*a for i in X] #y_pred=aX
plt.scatter(X, y)
plt.plot(X, y_pred, 'r') #? 작성
plt.xlabel("fuel")
plt.ylabel("distance")
plt.show()
```

```python
# 연료에 대한 거리를 예측하는 함수 정의
# 예측 모델 함수 정의: 주어진 연료(X)와 가중치(w)로 거리(y)를 추정
# ==========================
# 예측 함수 Ŷ = w * X
# ==========================
def pred_func(X, w):
    '''y=aX 모델에서 a는 가중치(w)이다.'''
    return [i*w for i in X]
```

```python
# 정의된 모델 함수를 사용하여 예측값(y_pred) 계산
a = 0.5

y_pred = pred_func(X, a) #? 작성
print(y_pred)
```

```python
plt.scatter(X, y)
plt.plot(X, y_pred, "r") #? 작성
plt.xlabel("fuel")
plt.ylabel("distance")
plt.show()
```

#Numpy(넘파이) 사용

## NumPy(Numerical Python)
- np라는 별칭을 사용하여 라이브러리를 불러 온다.
- 선형대수 기반의 프로그램을 쉽게 만들 수 있도록 지원하는 패키지이다.
- 정수나 실수와 같은 수치 데이터의 계산을 위해 사용되는 라이브러리이다.
- 루프를 사용하지 않고 대량 데이터의 배열 연산을 가능하게 하므로 빠른 배열 연산 속도를 보장한다.
- 많은 머신러닝 알고리즘, 데이터 분석이나 과학 계산을 위한 라이브러리가 numpy를 기반으로 작성되어 있다.

```python
import numpy as np
#연료에 대한 자동차의 이동 거리 데이터
car = [[ 4,  6],
       [ 8,  5],
       [12, 15],
       [16, 12],
       [20, 20]]

car = np.array(car)
print(car)
```

```python
#데이터 추출
X = car[:, 0] # 모든 행의 0번째 열
y = car[:, 1] #? 작성
print(X)
print(y)
```

```python
# 예측 함수 Ŷ = w * X
def pred_func(X, w):
    return w * X
# X는 넘파이 배열
# 벡터화 연산: 배열(벡터, 행렬 등)에 대해 반복문 없이 전체 데이터를 한 번에 처리하는 연산 방식
```

```python
# 정의된 모델 함수를 사용하여 예측값(y_pred) 계산
a = 0.9 #? 작성

y_pred = pred_func(X, a) #? 작성
print(y_pred)
```

```python
plt.scatter(X, y) #? 작성
plt.plot(X, y_pred, "r") #? 작성
plt.xlabel("fuel")
plt.ylabel("distance")
plt.show()
```
